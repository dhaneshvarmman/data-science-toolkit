```python
#------------Tutorial 02-------------------

'''
1. Create a numpy array of size 2 x 2
'''

import numpy as np
array_2x2 = np.array([
    [1,2],
    [3,4]
])
print("This is my numpy 2x2 array\n")
print(array_2x2)
```

```python
'''
2. Create a random numpy array of size 3 x 4
'''
random3x4 = np.random.rand(3,4)

print("Random 3x4 array\n")
print(random3x4)
```

```python
'''
3. Create a numpy array of ones and a numpy array of zeroes
'''
ones_array = np.ones((3,3))
zeros_array = np.zeros((3,3))

print("Array of Ones\n")
print(ones_array)

print("\nArray of Zeros\n")
print(zeros_array)
```

```python
'''
4. Create a random numpy array of size 6x3 and transpose it
'''
random6x3 = np.random.rand(6,3)
transposed6x3 = random6x3.T

print("Random 6x3 array\n")
print(random6x3)

print("\nTransposed 6x3 array\n")
print(transposed6x3)
```

```python
'''
5. Create two random numpy array of size n x p and p x m and then compute
   the dot product
'''
n = int(input("Enter dimension n: "))
m = int(input("Enter dimension m: "))
p = int(input("Enter dimension p: "))

random_nxp = np.random.rand(n,p)
random_pxm = np.random.rand(p,m)

dotproduct = np.dot(random_nxp,random_pxm)

print("\nDot product of two random dimensional arrays\n")
print(dotproduct)
```
